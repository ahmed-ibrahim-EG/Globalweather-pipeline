```python
import pyodbc
import pandas as pd
import logging
```

```python
# connect to SQL Server
conn = pyodbc.connect(driver ="{ODBC Driver 17 for SQL Server}", server = r"localhost\SQLEXPRESS", database = "DEPI", trusted_connection='yes')
# create a cursor (Query)
cursor = conn.cursor()
# create a table
table_sql = """
IF OBJECT_ID(N'dbo.reminders', N'U') IS NULL
BEGIN
    CREATE TABLE dbo.reminders (
        id INT IDENTITY(1,1) PRIMARY KEY,
        task_title NVARCHAR(255) NOT NULL,
        priority NVARCHAR(50) NOT NULL,
        status NVARCHAR(50) NOT NULL,
        due_date DATE NOT NULL
    );
END
"""

cursor.execute(table_sql)
conn.commit();
```

```python
# mock database
sample_tasks = [
    ("Build Data Pipeline", "High", "Pending", "2026-09-25"),
    ("Review SQL Window Functions", "Medium", "Completed", "2026-09-22"),
    ("Flask GET Endpoint Practice", "High", "Pending", "2026-09-23"),
    ("Update Documentation", "Low", "Pending", "2026-09-28"),
]
# insert into database
cursor.executemany(
    """
    INSERT INTO reminders (task_title, priority, status, due_date)
    VALUES (?, ?, ?, ?);
""",
    sample_tasks,
)
# save changes to database
conn.commit()
#close connection
conn.close()
# print success message
logging.info("Successfully inserted sample tasks into database.")
```
